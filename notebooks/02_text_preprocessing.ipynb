```python
import pandas as pd
import numpy as np
import re
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("../data/IMDB Dataset.csv")

df = df.drop_duplicates()

print("Dataset shape:", df.shape)
```

```python
def clean_html(text):
    text = re.sub(r"<.*?>", " ", text)
    text = re.sub(r"\s+", " ", text)
    return text.strip()
```

```python
df["cleaned_review"] = df["review"].apply(clean_html)
```

```python
df["cleaned_review"] = df["cleaned_review"].str.lower()
```

```python
def remove_special_characters(text):
    text = re.sub(r"[^a-z0-9\s]", " ", text)
    text = re.sub(r"\s+", " ", text)
    return text.strip()
```

```python
df["cleaned_review"] = df["cleaned_review"].apply(
    remove_special_characters
)

print("Text preprocessing completed!")
```

```python
special_characters = df["cleaned_review"].str.contains(
    r"[^a-z0-9\s]",
    regex=True
).sum()

print("Reviews containing special characters:", special_characters)

print("\nSample cleaned review:")
print(df["cleaned_review"][1][:500])
```

```python
print("Final dataset shape:", df.shape)

print("\nMissing values:")
print(df[["cleaned_review", "sentiment"]].isnull().sum())

print("\nSentiment distribution:")
print(df["sentiment"].value_counts())
```
